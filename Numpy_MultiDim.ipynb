```python
import numpy as np
```

```python
#Mais de uma dimensao
    #Criando um array bidimensional
mtz2_dim = np.array([[7,2,3], [5,6,7], [20,12,40]])
print(mtz2_dim)
```

```python
#criar arrays vazios tipificados
#empty significa que não são inicializados, não que são vazios
vazio = np.empty([3,2], dtype= int)
print(vazio)
print("--------")

#Criando uma matriz 4x3 com valores zeros
zeros = np.zeros([4,3])
print(zeros)
print("--------")

#valores igual a um
um = np.ones([4,3])
print(um)
print("--------")

#Matriz quadrada com diagonal principal com valor 1 e o resto 0
diagonal = np.eye(5)
print(diagonal)
```

```python
#valores aleatorios entre zero e um
alea = np.random.random((5))
print(alea)
print("--------")

#Valores aleatorios de uma distribuicao normal contendo negativos
alea2 = np.random.randn((5))
print(alea2)
print("--------")

#valores aleatorios 3 x 4
alea3 = (10*np.random.random((3,4)))
print(alea3)
```

```python
#outra forma de gerar aleatorios
#uso de semente
gnr = np.random.default_rng(1)
alea5 = gnr.random(3)
print(alea5)

#gerar inteiros
alea6 = gnr.integers(10, size=(3, 4))
print(alea6)
```

```python
#unique remove repetiçoes
j = np.array([11, 11, 12, 12, 13, 13])
j = np.unique(j)
print(j)
```

```python
#funçoes especificas
#cria a matriz bidimensional k
k = np.array([[1,1,1], [2,2,2], [3,3,3]])

#mostra a matriz k
print(k)

#mostra um elemento especifico da matriz
print(k[0][1])

#mostra o tamanho das dimensoes da matriz k
print(k.shape)
```
