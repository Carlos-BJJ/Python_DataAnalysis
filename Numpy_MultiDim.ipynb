```python
import numpy as np
```

```python
#Mais de uma dimensao
    #Criando um array bidimensional
mtz2_dim = np.array([[7,2,3], [5,6,7], [20,12,40]])
print(mtz2_dim)
```

```python
#criar arrays vazios tipificados
#empty significa que não são inicializados, não que são vazios
vazio = np.empty([3,2], dtype= int)
print(vazio)
print("--------")

#Criando uma matriz 4x3 com valores zeros
zeros = np.zeros([4,3])
print(zeros)
print("--------")

#valores igual a um
um = np.ones([4,3])
print(um)
print("--------")

#Matriz quadrada com diagonal principal com valor 1 e o resto 0
diagonal = np.eye(5)
print(diagonal)
```

```python
#valores aleatorios entre zero e um
alea = np.random.random((5))
print(alea)
print("--------")

#Valores aleatorios de uma distribuicao normal contendo negativos
alea2 = np.random.randn((5))
print(alea2)
print("--------")

#valores aleatorios 3 x 4
alea3 = (10*np.random.random((3,4)))
print(alea3)
```

```python
#outra forma de gerar aleatorios
#uso de semente
gnr = np.random.default_rng(1)
alea5 = gnr.random(3)
print(alea5)

#gerar inteiros
alea6 = gnr.integers(10, size=(3, 4))
print(alea6)
```

```python
#unique remove repetiçoes
j = np.array([11, 11, 12, 12, 13, 13])
j = np.unique(j)
print(j)
```

```python
#funçoes especificas
#cria a matriz bidimensional k
k = np.array([[1,1,1], [2,2,2], [3,3,3]])

#mostra a matriz k
print(k)

#mostra um elemento especifico da matriz
print(k[0][1])

#mostra o tamanho das dimensoes da matriz k
print(k.shape)
```

```python
#Funçoes Matematicas
#Mostra o maior valor da matriz k
print(k.max())

#Mostra o menor valor da matriz k
print(k.min())

#Mostra a soma dos valores da matriz k
print(k.sum())

#Mostra a média dos valores da matriz k
print(k.mean())

#Mostra o valor do desvio padrão (standard deviation) dos valores da matriz k
print(k.std())
```

```python
#Funções Universais, aplicadas a todos elementos
#Mostra o valor da raiz quadrada de todos elementos

k1 = np.array([1, 4, 9, 16, 25, 36])
print(np.sqrt(k1))

#Mostra o valor do exponecial de todos os elementos
print(np.exp(k1))
```

```python
#Extração de elementos
m = np.array([1, 2, 3, 4, 5, 6])

#Mostra o elemento da posição 2
print(m[1])
print("--------")

#Mostra o array criado a partir da posição 0, dois elementos
print(m[0:2])
print("--------")

#Mostra o array criado a partir da 2a posição
print(m[1:])
print("--------")

#Mostra o array criado a partir do antepenúltimo posição até o final
print(m[-3:])
```

```python
#Extração de linhas e colunas
l = np.array ([[4, 5], [6, 1], [7, 4]])
print(l)
print("--------")

#Primeira linha, todas as colunas
l_linha_1 = l[0, :]
print(l_linha_1)
print("--------")

#Segunda linha
l_linha_2 = l[1, :]
print(l_linha_2)
print("--------")

#todas as linhas, primeira coluna
l_coluna_1 = l[:, 0]
print(l_coluna_1)
print("--------")

#todas as linhas, segunda coluna
l_coluna_2 = l[:, 1]
print(l_coluna_2)
print("--------")
```

```python
#Adição e multiplicação de matrizes
n = np.array([[1, 2], [3, 4]])
o = np.array([[1, 1], [1, 1]])
res1 = n+o
print(res1)
print("--------")

res2 = n*o
print(res2)
print("--------")

p = np.array([[1, 2], [3, 4], [5, 6]])
q = np.array([[2, 1]])

print(p+q)
```

```python
#Transposição, rearranja um conjunto de 15 elementos de 0 a 14
#em 3 linhas e 5 colunas
f = np.arange(15).reshape((3,5))

#Mostra a matriz transposta entre linha e coluna
print(f)
print("--------")
s = f.T
print(s)
```

```python
#Outra forma de fazer mesmo resultado
r = np.arange(15).reshape((3,5))

print(r)
print("--------")

#rearranja um conjunto de 15 elementos
#mostra a matriz transposta entre linha e coluna
s = r.transpose((1,0))
print(s)
```
