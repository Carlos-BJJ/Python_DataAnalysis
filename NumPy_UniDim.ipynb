```python
import numpy as np
```

```python
#cria uma matriz unidimensional
mtz = np.array([12,13,14,15])
print(mtz)
print(type(mtz))
```

```python
#criando um array com tipo especifico
    #cria um array como float de 64 bits
mtzfloat = np.array([1,2,3], dtype= np.float64)
print(mtzfloat)
print(type(mtzfloat))

    #cria um array como int de 32 bits
mtzint = np.array([4,5,6], dtype= np.int32)
print(mtzint)
print(type(mtzint))
```

```python
#Mudando o tipo de array
mtznew = np.array([1, 2, 3, 4.5, 3.333333])
mtznewint = mtznew.astype(np.int32)
print(mtznew)
print(mtznewint)
```
